# 3 opening a csv file from an url

```python
import requests
from io import StringIO
u1 = "https://raw.githubusercontent.com/SLPeoples/Machine-Learning-A-Z/master/Part%2003%20-%20Classification/16_SVM/Social_Network_Ads.csv"
headers = {"User-Agent": "Mozilla/5.0"}
req = requests.get(u1, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

# 13 Handling Dates

```python
pd.read_csv('ipl_matches_data.csv',parse_dates=['city'])
```

```python
def rename(venue):
    if venue=="M Chinnaswamy Stadium":
        return "MCS"
    else:
        return venue
```

```python
rename("M Chinnaswamy Stadium")
```

# 14 convertors

```python
pd.read_csv('ipl_matches_data.csv',converters={'venue':rename})
```
