```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('train.csv')
```

# 1. Categorical Data

### a. Countplot

```python
sns.countplot(data=df, x='Embarked')
#df['Survived'].value_counts().plot(kind='bar')
```

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
sns.distplot(df['Age'])
```

```python
sns.boxplot(data=df, x='Age')
```
