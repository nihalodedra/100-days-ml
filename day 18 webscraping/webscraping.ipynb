```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

final = pd.DataFrame()

for j in range(1, 11):

    url = 'https://www.ambitionbox.com/list-of-companies?page={}'.format(j)
    headers = {'User-Agent': 'Mozilla/5.0'}
    webpage = requests.get(url, headers=headers).text
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div', class_='companyCardWrapper')

    name = []
    rating = []
    review_count = []
    ctype = []
    hq = []
    salaries = []
    interviews = []

    for i in company:
        name.append(i.find('meta', itemprop='name')['content'])

        # Rating
        r = i.find('div', class_='companyCardWrapper__ratingWrapper')

        if r:
            rating.append(r.text.strip().split()[0])
        else:
            rating.append('N/A')

        # Reviews, salaries, interviews
        actions = i.find_all('span',class_='companyCardWrapper__ActionCount')
        review_count.append(actions[0].text.strip() if len(actions) > 0 else '0')

        salaries.append(actions[1].text.strip() if len(actions) > 1 else '0')

        interviews.append(actions[2].text.strip() if len(actions) > 2 else '0')

        # Company type and headquarters
        info = i.find_all('span',class_='companyCardWrapper__interLinking')

        if len(info) > 0:

            details = info[0].text.split('|')
            ctype.append(details[0].strip())
            hq.append(
                details[1].strip()
                if len(details) > 1
                else 'N/A')

        else:
            ctype.append('N/A')
            hq.append('N/A')

    d = {
        'name': name,
        'rating': rating,
        'review_count': review_count,
        'company_type': ctype,
        'headquarters': hq,
        'salaries_count': salaries,
        'interview_count': interviews
    }

    df = pd.DataFrame(d)

    final = pd.concat([final, df],ignore_index=True
    )

```
